```python
import requests
import sys

url = "https://influencer-search.p.rapidapi.com/api/v1/influencer/search"
script_argument = sys.argv[1]
payload = { "query": { "keyword": script_argument } }
headers = {
	"content-type": "application/json",
	"X-RapidAPI-Key": "a619bfcad9msh56f2f66563cadb4p186d94jsn0a59d4ce6269",
	"X-RapidAPI-Host": "influencer-search.p.rapidapi.com"
}

response = requests.post(url, json=payload, headers=headers)

content= response.content
import json 
import pandas as pd 
from pandas import json_normalize 
dict_data= json.loads(content)
df_influencer = json_normalize(dict_data['data'])
df_influencer['location'].fillna('',inplace=True)
df_influencer['time_zone'].fillna('',inplace=True)
df_influencer['lang'].fillna('',inplace=True)
df_influencer['url'].fillna('',inplace=True)
df_influencer['utc_offset'].fillna('',inplace=True)
df_influencer['profile_use_background_image'].fillna('',inplace=True)
df_influencer['protected'].fillna('',inplace=True)
df_influencer['geo_enabled'].fillna('',inplace=True)
import pymongo
con = pymongo.MongoClient("mongodb://localhost:27017/")
db = con.DCSTEC
collection = db.Influencers
data = df_influencer.to_dict(orient='records')
collection.insert_many(data)

print(json.dumps(dict_data, indent=2))
```
