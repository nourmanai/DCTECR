```python
import requests

# Set your Bitly access token
access_token = "77e0ea67635a04f6950b97a711a7044cbbc5d18d"

# Define the endpoint URL for creating a bitlink
endpoint = "https://api-ssl.bitly.com/v4/shorten"
# Set the headers with the access token
headers = {
    "Authorization": f"Bearer 77e0ea67635a04f6950b97a711a7044cbbc5d18d",
    "Content-Type": "application/json"
}

# Define the long URL to shorten
long_url = "https://www.google.com/webhp?hl=fr&sa=X&ved=0ahUKEwi21q2c-YiAAxXCXvEDHaC4Cu8QPAgI"

# Create the request payload
payload = {
    "long_url": long_url
}

# Send a POST request to the Bitly API
response = requests.post(endpoint, headers=headers, json=payload)

# Check if the request was successful (HTTP status code 200 or 201)
if response.status_code in (200, 201):
    # Parse the response JSON
    data = response.json()
    # Extract the shortened URL (bitlink) from the response
    short_url = data.get("link")
    print("Shortened URL:", short_url)
else:
    print("Error:", response.status_code)
```

```python
import requests

# Set your Bitly access token
access_token = "77e0ea67635a04f6950b97a711a7044cbbc5d18d"

# Define the endpoint URL for retrieving click data
bitlink_id = "bit.ly/3JUQJWK"  # Replace with your actual bitlink ID
endpoint = f"https://api-ssl.bitly.com/v4/bitlinks/{bitlink_id}/clicks"

# Set the headers with the access token
headers = {
    "Authorization": f"Bearer 77e0ea67635a04f6950b97a711a7044cbbc5d18d",
    "Content-Type": "application/json"
}

# Set the parameters for retrieving click data
params = {
    "unit": "month",
    "units": 1,
    "unit_reference": "2006-01-02T15:04:05-0700"
}

# Send a GET request to the Bitly API
response = requests.get(endpoint, headers=headers, params=params)

# Check if the request was successful (HTTP status code 200)
if response.status_code == 200:
    # Parse the response JSON
    data = response.json()
    # Extract the click data from the response
    click_data = data.get("link_clicks")
    print("Click data:", click_data)
else:
    print("Error:", response.status_code)
```
