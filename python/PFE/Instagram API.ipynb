```python
import requests
import json
from bs4 import BeautifulSoup
import sys

# Set the endpoint URL
url = "https://www.instagram.com/{username}/?__a=1"

# Set the Instagram username
username = sys.argv[1]
username = "manainoor"
# Send a GET request to the API
response = requests.get(url.format(username=username))
print (response)
# Check if the request was successful (HTTP status code 200)
if response.status_code == 200:
    try:
        # Attempt to parse the response as JSON
        html_content = response.content.decode('utf-8')
        soup = BeautifulSoup(html_content, 'html.parser')

        # Now you can use BeautifulSoup to navigate and extract information from the HTML
        # For example, print the text content of all paragraphs:
        links = soup.find_all('link')
        linksF = []
        for link in links:
            linksF.append(link.get('href'))
        
        print(linksF)

    except json.JSONDecodeError:
        # If parsing as JSON fails, handle HTML content separately
        html_content = response.content.decode('utf-8')
        print ("exception")
else:
    # Handle non-200 status code
    print(f"Error: {response.status_code}")
```
