```python
import requests

url = "https://newsnow.p.rapidapi.com/"

payload = {
	"text": "Top news",
	"region": "wt-wt"
}
headers = {
	"content-type": "application/json",
	"X-RapidAPI-Key": "a619bfcad9msh56f2f66563cadb4p186d94jsn0a59d4ce6269",
	"X-RapidAPI-Host": "newsnow.p.rapidapi.com"
}

response = requests.post(url, json=payload, headers=headers)

print(response.json())
```

```python
contentTopNews= response.content
```

```python
import json 
import pandas as pd 
from pandas import json_normalize 
import pymongo
con = pymongo.MongoClient("mongodb://localhost:27017/")
db = con.DCSTEC
```

```python
dict= json.loads(contentTopNews)

df_TopNews = json_normalize(dict['news'])
```

```python
df_TopNews['image'].fillna('',inplace=True)
```

```python
collection = db.TopNews
data = df_TopNews.to_dict(orient='records')
collection.insert_many(data)
```
